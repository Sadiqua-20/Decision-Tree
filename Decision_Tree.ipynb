# **THEORETICAL QUESTIONS**

## Q1. What is a Decision Tree, and how does it work?

## ANS 1.

A **Decision Tree** is a type of supervised machine learning algorithm used for both classification and regression tasks. It works by learning simple decision rules from data features to make predictions.


### **How It Works**

1. The algorithm begins at the top of the tree, called the **root**, which represents the entire dataset.

2. At each node, the algorithm selects the feature and threshold that best split the data into subgroups with the most uniform target values (i.e., maximizing information gain or minimizing impurity).

3.   Each internal node represents a **test** on a feature (e.g., "Is Age < 30?"). Based on the outcome (yes/no), the data moves down to a **child node**.

4. When the data can't be split further (or a stopping criterion is met), the node becomes a leaf and represents a prediction.

## Q2. What are impurity measures in Decision Trees?


## ANS 2.

**Impurity measures** in Decision Trees evaluate how mixed or pure the target values are at a node. Common ones include:

- **Gini Impurity**: Measures the probability of incorrectly classifying a randomly chosen element.
- **Entropy**: Measures the amount of information (or uncertainty) in the node.
- **Variance**: Used in regression trees to measure the spread of target values.

Lower impurity means a more "pure" node (i.e., mostly one class).

## Q3. What is the mathematical formula for Gini Impurity?

## ANS 3.

The mathematical formula for Gini Impurity is:

$Gini = 1 - \sum_{i=1}^{n} p_i^2$  

where,
- $n$  = number of classes  
- $p_i$  = proportion of samples belonging to class $i$ in the node

## Q4. What is the mathematical formula for Entropy?

## ANS 4.

The mathematical formula for Entropy is:

$Entropy = - \sum_{i=1}^{n} p_i \log_2(p_i)$


Where:
- $n$ = number of classes  
- $p_i$ = proportion of samples in class $i$

## Q5. What is Information Gain, and how is it used in Decision Trees?


## ANS 5.

**Information Gain** measures the reduction in impurity (like Entropy or Gini) after a dataset is split on a feature. It is used in Decision Trees to choose the best feature for splitting at each node.


$\text{Information Gain} = \text{Impurity (parent)} - \sum_{k} \frac{N_k}{N} \times \text{Impurity (child}_k\text{)}$

where,
- $N_k$  = number of samples in child $k$
- $N$ = total number of samples in the parent node

Higher Information Gain means a better split.

## Q6. What is the difference between Gini Impurity and Entropy?


## ANS 6.

The differences between Gini Impurity and Entropy are as follows:

1. **Formula**:  
   - Gini: $ 1 - \sum_{i=1}^{n} p_i^2$  
  
   - Entropy: $- \sum_{i=1}^{n} p_i \log_2(p_i)$


2. **Concept**:  
   - Gini measures the probability of misclassifying a random sample.  
   - Entropy measures the level of disorder or uncertainty.

3. **Range (Binary classification)**:  
   - Gini: 0 (pure) to 0.5  
   - Entropy: 0 (pure) to 1

4. **Computation**:  
   - Gini is faster (no logarithms).  
   - Entropy is more computationally intensive (uses logarithms).

5. **Splitting Behavior**:  
   - Gini tends to split on the most frequent class.  
   - Entropy is more sensitive to class distribution.

## Q7. What is the mathematical explanation behind Decision Trees?


## ANS 7.


1. **Split Data**: Choose a feature and threshold that best splits the data into subsets.

2. **Calculate Impurity**:
   - Use **Gini** or **Entropy** to measure how mixed the classes are

3. **Information Gain**:
   - Choose the split that **maximizes** Information Gain:  
  $\text{Information Gain} = \text{Impurity (parent)} - \sum_{k} \frac{N_k}{N} \times \text{Impurity (child}_k\text{)}$


4. **Repeat**: Recursively split until stopping conditions are met (e.g., pure nodes or max depth).

5. **Predict**: Leaf nodes output the **majority class** (classification) or **average value** (regression).

## Q8. What is Pre-Pruning in Decision Trees?


## ANS 8.

**Pre-pruning** (also called **early stopping**) is a technique used to prevent a Decision Tree from growing too deep and to prevent overfitting the training data.

- Stop the tree from splitting further during training if a certain condition is met.

- We do hyperparameter tuning to select the best parameters

### **Purpose**:
Improves **generalization** by avoiding overly complex trees.

## Q9. What is Post-Pruning in Decision Trees?


## ANS 9.

**Post-pruning** (also called **pruning after training**) is a technique where a fully grown Decision Tree is simplified by removing branches that add little predictive power.


### **How It Works**:
- Grow the decision tree completely.
- Evaluate subtrees using a validation set.
- Remove branches (cut back parts of it) that don't improve accuracy significantly, to reduce overfitting..

### **Purpose**:  
Improves **generalization** and reduces model complexity.



## Q10. What is the difference between Pre-Pruning and Post-Pruning?


## ANS 10.

The differenceS between **Pre-Pruning** and **Post-Pruning** in Decision Trees are as follows:


### **Pre-Pruning (Early Stopping)**

1. **Applied during tree construction** — the tree stops growing when a condition is met.  
2. Prevents overfitting early by limiting tree size in advance.  
3. Conditions include max depth, min samples per node, or min information gain.  
4. Faster training since the tree is never fully built.  
5. Risk of underfitting if stopped too soon.  
6. Does not require a validation set (but can use one).


### **Post-Pruning (Pruning After Training)**

1. Applied after the full tree is built — removes unnecessary branches.  
2. Simplifies an overfit tree by trimming parts that add little value.  
3. Uses validation data or techniques.  
4. Slower training as the full tree is created first.  
5. Better generalization in many cases.  
6. Reduces overfitting while maintaining accuracy.



## Q11. What is a Decision Tree Regressor?


## ANS 11.

A **Decision Tree Regressor** from `sklearn.tree` is a machine learning model used to predict **continuous numeric values** by learning simple decision rules from input features.


### **Definition**:
```
from sklearn.tree import DecisionTreeRegressor
```
This class is part of **scikit-learn**, and it builds a tree where each split minimizes a regression loss (like Mean Squared Error).


### **How It Works**:
- Splits data into regions based on features.
- At each leaf node, it predicts the **average target value** of the training samples in that region.



## Q12. What are the advantages and disadvantages of Decision Trees	?

## ANS 12.

The **advantages and disadvantages of Decision Trees** are as follows:


### **Advantages**:

1. **Easy to understand and interpret** – visually intuitive tree structure.  
2. No need for feature scaling or normalization.  
3. Can handle both numerical and categorical data.  
4. Works well with nonlinear relationships.  
5. Requires little data preprocessing (e.g., no need to remove missing values in some implementations).  
6. Fast training and prediction for small to medium datasets.


### **Disadvantages**:

1. Prone to overfitting, especially with deep trees.  
2. **Unstable** – small changes in data can lead to a completely different tree.  
3. **Greedy splitting** – may not always find the global optimal solution.  
4. Biased toward features with more levels (for categorical data).  
5. Lower accuracy compared to ensemble methods like Random Forest or Gradient Boosting.  



## Q13. How does a Decision Tree handle missing values?


## ANS 13.

**Decision Trees** can handle **missing values** in a few different ways, depending on the implementation:


**1. Ignoring During Split (Simple Skipping)**  
- Some implementations **ignore** rows with missing values for a feature when calculating the best split on that feature.

**2. Surrogate Splits**  
- When a primary split feature has a missing value, the tree uses a **surrogate feature**—another feature that closely mimics the primary split.

**3. Assigning to the Most Common Branch**  
- When a value is missing, the algorithm may assign the sample to the **branch** with the **majority of similar samples**.

**4. Imputation Before Training (Manual Step)**  
- In `scikit-learn`, Decision Trees **don’t natively handle missing values**, so we typically use **imputation** (e.g., with `SimpleImputer`) to fill in missing values before fitting the model.



## Q14. How does a Decision Tree handle categorical features?


## ANS 14.

Decision Trees handle categorical features by:

1. **Splitting on categories**: The tree splits data based on specific categories or values of the categorical feature.
2. **Choosing best splits**: The algorithm selects the best categorical feature to split the data, often based on metrics like Gini impurity or entropy.

Some Decision Tree implementations use techniques like:

1. One-hot encoding
2. Label encoding
3. Direct categorical splits

The goal is to find the optimal split that separates the data effectively for classification or regression tasks.

## Q27. Write a Python program to train a Decision Tree Classifier, apply Cost Complexity Pruning (CCP), and visualize its effect on accuracy.

````python
from sklearn.datasets import load_iris
iris = load_iris()
x27  = iris.data
y27 = iris.target

x27_train, x27_test, y27_train, y27_test = train_test_split(x27, y27, test_size=0.2, random_state=1)

clf = DecisionTreeClassifier()
path = clf.cost_complexity_pruning_path(x27_train, y27_train)
ccp_alphas = path.ccp_alphas

clfs = []
for alpha in ccp_alphas:
    clf = DecisionTreeClassifier(random_state=1, ccp_alpha=alpha)
    clf.fit(x27_train, y27_train)
    clfs.append(clf)

test_accuracies = [accuracy_score(y27_test, clf.predict(x27_test)) for clf in clfs]

plt.figure(figsize=(8, 5))
plt.plot(ccp_alphas, test_accuracies, marker='o')
plt.xlabel("ccp_alpha (pruning strength)")
plt.ylabel("Test Accuracy")
plt.title("Effect of Cost Complexity Pruning on Test Accuracy")
plt.grid(True)
plt.show()
````
